Clean the Dataset

```python
print("Missing values")
print(df.isnull().sum())
```

```python
print("Duplicate records")
print(df.duplicated().sum())
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
```

```python
print("Rows and columns after cleaning:", df.shape)

print("Missing values after cleaning:")
print(df.isnull().sum())

print("Duplicates after cleaning:")
print(df.duplicated().sum())

display(df.head())
```

```python
print(df['High'].max())
```

```python
print(df['Low'].min())
```

```python
df['Daily_Delta'] = df['Close'] - df['Open']
```

```python
display(df[['Date', 'Open', 'Close', 'Daily_Delta']])
```

```python
df['Daily_return'] = (df['Close'] - df['Open']) / df['Open']*100
```

```python
display(df[['Date', 'Open', 'Close', 'Daily_Delta','Daily_return']])
```

```python
print(df[['Date', 'Volume']].head())
```

```python
df['Volume_MA_7'] = df['Volume'].rolling(7).mean()
```

```python
volume_mean = df['Volume'].mean()
volume_std = df['Volume'].std()
```

```python
df['Volume_Anomaly'] = df['Volume'] > (volume_mean + 2 * volume_std)
```

```python
anomalous_days = df[df['Volume_Anomaly'] == True]

print("Anomalous Trading Days:")
print(anomalous_days[['Date', 'Volume']])
```

```python
mean_return = df['Daily_return'].mean()

print("Mean Return:", mean_return)
```

```python
variance_return = df['Daily_return'].var()

print("Variance:", variance_return)
```

```python
std_return = df['Daily_return'].std()

print("Standard Deviation:", std_return)
```
